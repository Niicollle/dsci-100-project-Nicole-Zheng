#### Task 3.1: Tidy the data (5 points)

```python
# Each row represents one school’s tournament appearance in one year.
# Retain identifiers, the response, pre-tournament predictors, and a categorical variable.
columns = ["year", "school", "seed", "reg_wins_pct", "conf_wins_pct", "bid", "tourney_wins"]
basketball = basketball_raw.loc[:, columns].copy()

# Convert numeric fields explicitly; invalid entries become missing values.
numeric_columns = ["year", "seed", "reg_wins_pct", "conf_wins_pct", "tourney_wins"]
for column in numeric_columns:
    basketball[column] = pd.to_numeric(basketball[column], errors="coerce")

# Clean category labels while preserving missing values.
basketball["bid"] = basketball["bid"].astype("string").str.strip()
# A missing response cannot be used for supervised learning.
basketball = basketball.dropna(subset=["tourney_wins"]).reset_index(drop=True)
# Seed 0 denotes the historical opening-round designation, not the strongest seed.
basketball.loc[basketball["seed"] == 0, "seed"] = np.nan

# Split before exploration. Keep the test set untouched until final evaluation.
from sklearn.model_selection import train_test_split
train_data, test_data = train_test_split(basketball, test_size=0.25, random_state=SEED)
display(basketball.head(6))
print("Training rows:", len(train_data), "Test rows:", len(test_data))
print("Missing predictors are retained and will be imputed inside cross-validation.")
```

I will use k-nearest neighbours (KNN) regression with tournament seed, regular-season win percentage, and conference win percentage to predict tournament wins. These predictors are available before tournament outcomes occur; tournament-derived totals and finishing position are excluded to avoid outcome leakage. KNN is appropriate for a numeric response and can capture nonlinear relationships without specifying a linear equation. Its weaknesses include sensitivity to scaling and the choice of k, limited extrapolation, and possible fractional or smoothed predictions for an integer outcome.

I have randomly reserved 25% of appearances as a test set before exploratory analysis, with the remaining 75% used for model development. Within the training set, shuffled five-fold cross-validation will select k from 3, 5, 7, 9, 15, 25, and 51 using the lowest root mean squared error (RMSE). A pipeline will fit median imputation and standardization separately within each training fold, preventing validation leakage. The selected pipeline will then be refitted on all training observations and evaluated once on the held-out test set. I will compare test RMSE with a training-mean baseline and report MAE as a complementary measure. RMSE penalizes large errors; MAE describes typical absolute error in wins. Repeated schools and changes across historical eras limit independence and generalization to future tournaments; this random split evaluates held-out historical appearances rather than future-year forecasting.

#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.dummy import DummyRegressor
from sklearn.model_selection import GridSearchCV, KFold
from sklearn.metrics import mean_squared_error, mean_absolute_error

predictors = ["seed", "reg_wins_pct", "conf_wins_pct"]
X_train, y_train = train_data[predictors], train_data["tourney_wins"]
X_test, y_test = test_data[predictors], test_data["tourney_wins"]

# Fit preprocessing only on each training fold, never on validation or test data.
knn_pipeline = make_pipeline(SimpleImputer(strategy="median"), StandardScaler(), KNeighborsRegressor())
folds = KFold(n_splits=5, shuffle=True, random_state=SEED)
search = GridSearchCV(
    knn_pipeline,
    {"kneighborsregressor__n_neighbors": [3, 5, 7, 9, 15, 25, 51]},
    cv=folds, scoring="neg_root_mean_squared_error", refit=True
)
search.fit(X_train, y_train)

# Report training cross-validation results before evaluating the test set.
cv_results = pd.DataFrame({
    "k": search.cv_results_["param_kneighborsregressor__n_neighbors"],
    "mean_validation_RMSE": -search.cv_results_["mean_test_score"],
    "fold_RMSE_SD": search.cv_results_["std_test_score"]
}).sort_values("k")
display(cv_results)
print("Selected k:", search.best_params_["kneighborsregressor__n_neighbors"])

# Evaluate the selected model and a training-mean baseline on the same test set.
test_predictions = search.predict(X_test)
baseline = DummyRegressor(strategy="mean").fit(X_train, y_train)
baseline_predictions = baseline.predict(X_test)
metrics = pd.DataFrame({
    "model": ["KNN regression", "Training-mean baseline"],
    "test_RMSE": [np.sqrt(mean_squared_error(y_test, test_predictions)), np.sqrt(mean_squared_error(y_test, baseline_predictions))],
    "test_MAE": [mean_absolute_error(y_test, test_predictions), mean_absolute_error(y_test, baseline_predictions)]
})
display(metrics)
print("Relative RMSE reduction:", 1 - metrics.loc[0, "test_RMSE"] / metrics.loc[1, "test_RMSE"])
```

The table below compares the selected KNN model and a baseline that always predicts the mean training-set tournament wins. Both are evaluated on the same held-out observations, so smaller RMSE and MAE indicate better prediction. RMSE is measured in tournament wins and emphasizes large misses. The cross-validation curve shows how validation RMSE changes with k; k is chosen using training cross-validation, not test performance. The error bars show variation across folds, not confidence intervals.

```python
# Present held-out performance, then visualize training-only model selection.
display(metrics.round(3))
cv_plot_data = cv_results.copy()
cv_plot_data["lower"] = cv_plot_data["mean_validation_RMSE"] - cv_plot_data["fold_RMSE_SD"]
cv_plot_data["upper"] = cv_plot_data["mean_validation_RMSE"] + cv_plot_data["fold_RMSE_SD"]
base = alt.Chart(cv_plot_data).encode(x=alt.X("k:Q", title="Number of neighbours (k)"))
curve = base.mark_line(point=True).encode(y=alt.Y("mean_validation_RMSE:Q", title="Cross-validation RMSE (wins)", scale=alt.Scale(zero=False)))
bars = base.mark_rule().encode(y="lower:Q", y2="upper:Q")
(curve + bars).properties(title="Choosing k using five-fold training cross-validation", width=550, height=320)
```
